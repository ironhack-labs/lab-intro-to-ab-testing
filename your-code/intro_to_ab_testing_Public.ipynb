```python
### Import Libraries
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

> Simulate User Data, We’ll assume:

- Version A: 1000 visitors, 150 clicks.

- Version B: 1050 visitors, 180 clicks.

```python
# Simulate data
np.random.seed(42)
visitors_a = 1000
clicks_a = 150
visitors_b = 1050
clicks_b = 180
```

```python
# Version A
data_a = [1]*150 + [0]*850

# Version B
data_b = [1]*180 + [0]*870

# Convert to arrays
data_a = np.array(data_a)
data_b = np.array(data_b)

print(len(data_a), "visitors in A")
print(len(data_b), "visitors in B")
```

📊 Perform A/B Test
We’ll use a Chi-Square Test to compare proportions.

- Hypotheses

```python
# Null Hypothesis (H₀): Version A clicks_rate = Version B clicks_rate.
# Alternative Hypothesis Version A clicks_rate != Version B clicks_rate.
```


- Run the Test

```python
# Contingency table
table = [[150, 850],
         [180, 870]]

# Chi-Square test
chi2, p, dof, expected = stats.chi2_contingency(table)

print("Chi-square statistic:", chi2)
print("p-value:", p)
```

```python
# failed to reject h_0
```

```python
ctr_a = 150 / 1000
ctr_b = 180 / 1050

versions = ['Version A', 'Version B']
ctrs = [ctr_a, ctr_b]

# Plot
plt.bar(versions, ctrs)
plt.ylabel("Click Through Rate (CTR)")
plt.title("CTR Comparison: Version A vs Version B")

plt.show()
```

#### **Objective**:

### **Problem Statement**  
Analyze the [Cookie Cats dataset](https://www.kaggle.com/datasets/yufengsui/mobile-games-ab-testing) to determine if moving the game’s first gate from level 30 (`gate_30`) to level 40 (`gate_40`) affects player retention.  

### **Tasks**

   - Compare **1-day retention rates** between `gate_30` and `gate_40` using a Z-test.  
2. **Sequential Testing**:  
   - Simulate checking results at intervals (every 500 users) to avoid "peeking bias."  
3. **Power Analysis**:  
   - Calculate the required sample size to detect a 2% difference in retention (α = 0.05, power = 0.8).  
4. **Optional**:

### 🛠 **Setup & Data Loading**  


```python
# Import Libraries
```

```python
# Load dataset
df1 = pd.read_csv("https://raw.githubusercontent.com/Dee-M123/lab-intro-to-ab-testing/refs/heads/main/your-code/data/cookie_cats.csv")  # Columns: userid, version, sum_gamerounds, retention_1, retention_7

df1.head()
```

📊 Part 1: Retention Rate Analysis
> Compare 1-Day Retention

```python
retention_gate30 = df1[df1['version'] == 'gate_30']['retention_1'].mean()
retention_gate40 = df1[df1['version'] == 'gate_40']['retention_1'].mean()

print("1-Day Retention (gate_30):", retention_gate30)
print("1-Day Retention (gate_40):", retention_gate40)
```

> Statistical Test

```python
from statsmodels.stats.proportion import proportions_ztest

# Number of retained players
ret_gate30 = df1[df1['version'] == 'gate_30']['retention_1'].sum()
ret_gate40 = df1[df1['version'] == 'gate_40']['retention_1'].sum()

# Total players
n_gate30 = df1[df1['version'] == 'gate_30']['retention_1'].count()
n_gate40 = df1[df1['version'] == 'gate_40']['retention_1'].count()

# Z-test
count = [ret_gate30, ret_gate40]
nobs = [n_gate30, n_gate40]

z_stat, p_value = proportions_ztest(count, nobs)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

📈 Part 2: Sequential Testing
>  Why Sequential Testing?

- Avoids "peeking" at results prematurely by checking at intervals.

```python
# Sequential Testing: check results every 500 users

from statsmodels.stats.proportion import proportions_ztest

step = 500

for i in range(step, len(df1), step):

    sample = df1.iloc[:i]

    ret30 = sample[sample['version'] == 'gate_30']['retention_1'].sum()
    n30 = sample[sample['version'] == 'gate_30']['retention_1'].count()

    ret40 = sample[sample['version'] == 'gate_40']['retention_1'].sum()
    n40 = sample[sample['version'] == 'gate_40']['retention_1'].count()

    count = [ret30, ret40]
    nobs = [n30, n40]

    z_stat, p_value = proportions_ztest(count, nobs)

    print(f"Users checked: {i}, p-value: {p_value}")
```

⚡ Part 3: Power Analysis
>  Determine Required Sample Size

```python

from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

# baseline retention (example from gate_30)
p1 = 0.448
p2 = 0.468  # detect 2% improvement

effect_size = proportion_effectsize(p1, p2)

analysis = NormalIndPower()
sample_size = analysis.solve_power(effect_size=effect_size, power=0.8, alpha=0.05)

print("sample size per group:", round(sample_size))
```
