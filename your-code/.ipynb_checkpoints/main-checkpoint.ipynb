#### 🛠 **Setup & Data Generation**  
First, let’s create synthetic data for the experiment.  


```python
### Import Libraries
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

> Simulate User Data, We’ll assume:

- Version A: 1000 visitors, 150 clicks.

- Version B: 1050 visitors, 180 clicks.

```python
# Simulate data
np.random.seed(42)
visitors_a = 1000
clicks_a = 150
visitors_b = 1050
clicks_b = 180
```

```python
# To Calculate Click-Through Rate, divide the total number of clicks by the total number of visitors. 
# To get the percentage, you need to multiply this number by 100. This is the rate of how many users both viewed and clicked your ad.
ctr_a = (clicks_a / visitors_a)*100
ctr_b = (clicks_b / visitors_b)*100
```

```python
ctr_a
```

```python
ctr_b
```

```python
visitors_A = 1000
clicks_A = 150
visitors_B = 1050
clicks_B = 180

non_clicks_A = visitors_A - clicks_A
non_clicks_B = visitors_B - clicks_B

data = pd.DataFrame({
    "Version":["A", "B"],
    "Visitors": ["visitors_A", "visitors_B"],
    "Clicks": ["clicks_A", "clicks_B"],
    "Non-Clicks": ["non_clicks_A", "non_clicks_B"]})

contingency_table = np.array([[clicks_A, non_clicks_A],[clicks_B, non_clicks_B]])
contingency_table
```

```python
data
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
p_value
```

📊 Perform A/B Test
We’ll use a Chi-Square Test to compare proportions.

- Hypotheses

```python
# Null Hypothesis (H₀): There is no difference between the CTRs of Version A and Version B
# Alternative Hypothesis (H1): There is a difference between the CTRs of Version A and Version B
```

```python
# Bar chart for Clicks
ctr_a = clicks_A / visitors_A
ctr_b = clicks_B / visitors_B

plt.bar(data["Version A", "Version B"], data["ctr_a", "ctr_b"], color=["skyblue", "orange"])
plt.ylabel("Number of Clicks")
plt.title("Clicks by Version")
plt.show()
```
